```python
# Data Manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display Settings
pd.set_option("display.max_columns", None)

# Style
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10,6)


# ============================================================
# File Upload
# Upload the earthquake dataset
# ============================================================

from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)
```

```python
# ============================================================
# Data Types
# Display the data type of each column
# ============================================================

df.dtypes
```

```python
# ============================================================
# Convert Date-Time Column
# Convert date_time column into datetime format
# ============================================================

df["date_time"] = pd.to_datetime(df["date_time"])
```

```python
# ============================================================
# Handle Missing Values
# Select numerical columns
# ============================================================

numeric_cols = df.select_dtypes(include=np.number).columns


# Fill missing numerical values with the median
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
# ============================================================
# Check Missing Values After Preprocessing
# ============================================================

df.isnull().sum()
```

```python
# ============================================================
# Correlation Heatmap
# Show correlation between numerical earthquake variables
# ============================================================

plt.figure(figsize=(15,10))

sns.heatmap(df.corr(numeric_only=True),
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
# ============================================================
# Magnitude Distribution
# Display the distribution of earthquake magnitudes
# ============================================================

sns.histplot(df["magnitude"],
             bins=30,
             kde=True)

plt.title("Magnitude Distribution")

plt.show()
```

```python
# ============================================================
# Tsunami Distribution
# Display the number of earthquakes associated
# with and without a tsunami
# ============================================================

plt.figure(figsize=(8,5))

sns.countplot(data=df,
              x="tsunami")

plt.xticks(rotation=45)

plt.title("Tsunami Distribution")

plt.show()
```

```python
# ============================================================
# Top 10 Countries
# Display the countries with the highest number
# of recorded earthquakes
# ============================================================

plt.figure(figsize=(12,6))

sns.countplot(data=df,
              y="country",
              order=df["country"].value_counts().index[:10])

plt.title("Top 10 Countries")

plt.show()
```

```python
# ============================================================
# Depth Distribution
# Display the distribution of earthquake depths
# ============================================================

sns.histplot(df["depth"],
             bins=30,
             kde=True)

plt.title("Depth Distribution")

plt.show()
```

```python
# ============================================================
# CDI Distribution
# Display the distribution of CDI values
# ============================================================

sns.histplot(df["cdi"],
             bins=30,
             kde=True)

plt.title("CDI Distribution")

plt.show()
```

```python
# ============================================================
# MMI Distribution
# Display the distribution of MMI values
# ============================================================

sns.histplot(df["mmi"],
             bins=30,
             kde=True)

plt.title("MMI Distribution")

plt.show()
```

```python
# ============================================================
# Significance Distribution
# Display the distribution of earthquake significance values
# ============================================================

sns.histplot(df["sig"],
             bins=30,
             kde=True)

plt.title("Significance Distribution")

plt.show()
```

```python
# ============================================================
# Magnitude Distribution
# Display the magnitude distribution again
# ============================================================

sns.histplot(df["magnitude"],
             bins=30,
             kde=True)

plt.title("Magnitude Distribution")

plt.show()
```

```python
# ============================================================
# Magnitude Outliers
# Use a boxplot to identify possible outliers
# ============================================================

plt.figure(figsize=(8,5))

sns.boxplot(x=df["magnitude"])

plt.title("Magnitude Outliers")

plt.show()
```

```python
# ============================================================
# Depth vs Magnitude
# Visualize the relationship between earthquake depth
# and earthquake magnitude
# ============================================================

sns.scatterplot(data=df,
                x="depth",
                y="magnitude")

plt.title("Depth vs Magnitude")

plt.show()
```

```python
# ============================================================
# Average Magnitude by Country
# Calculate and display the top 10 countries
# based on average earthquake magnitude
# ============================================================

country = df.groupby("country")["magnitude"].mean().sort_values(ascending=False).head(10)

country.plot(kind="bar")

plt.ylabel("Average Magnitude")

plt.title("Top 10 Countries by Average Magnitude")

plt.show()
```

```python
# ============================================================
# Pairplot
# Show relationships between selected earthquake variables
# ============================================================

sns.pairplot(df[["magnitude","depth","cdi","mmi","sig"]])

plt.show()
```

```python
# ============================================================
# Pairplot
# Display the same selected variables again
# ============================================================

sns.pairplot(df[["magnitude","depth","cdi","mmi","sig"]])

plt.show()
```
