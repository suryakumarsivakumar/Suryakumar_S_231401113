```python
# Data Manipulation and Visualization
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# File Upload
# Upload the earthquake CSV dataset
# ============================================================

from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)
```

```python
# ============================================================
# Display First 5 Rows
# Display the first five records of the dataset
# ============================================================

print(df.head())
```

```python
# ============================================================
# Dataset Shape
# Display the number of rows and columns
# ============================================================

print(df.shape)
```

```python
# ============================================================
# Data Types
# Display the data type of each column
# ============================================================

print(df.dtypes)
```

```python
# ============================================================
# Statistical Summary
# Display descriptive statistics of numerical columns
# ============================================================

print(df.describe())
```

```python
# ============================================================
# High Magnitude Earthquakes
# Filter earthquakes with magnitude greater than 6
# ============================================================

high_magnitude = df[df["magnitude"] > 6]

print(high_magnitude.head())
```

```python
# ============================================================
# Earthquakes in a Specific Country
# Filter the earthquake records for India
# ============================================================

india = df[df["country"] == "India"]

print(india.head())
```

```python
# ============================================================
# Sort Earthquakes by Magnitude
# Sort the dataset from highest to lowest magnitude
# ============================================================

sorted_df = df.sort_values(
    by="magnitude",
    ascending=False
)

print(sorted_df.head())
```

```python
# ============================================================
# Descriptive Statistics of Magnitude
# Calculate mean, median, mode and standard deviation
# ============================================================

print("Mean :", df["magnitude"].mean())

print("Median :", df["magnitude"].median())

print("Mode :")

print(df["magnitude"].mode())

print("Standard Deviation :", df["magnitude"].std())
```

```python
# ============================================================
# Average Magnitude by Country
# Calculate the average earthquake magnitude for each country
# ============================================================

country_avg = df.groupby("country")["magnitude"].mean()

print(country_avg.sort_values(ascending=False).head(10))
```

```python
# ============================================================
# Top 10 Countries by Average Magnitude
# Display the top 10 countries based on average magnitude
# ============================================================

country_avg.sort_values(ascending=False).head(10).plot(kind="bar")

plt.title("Top 10 Countries by Average Earthquake Magnitude")

plt.ylabel("Average Magnitude")

plt.show()
```

```python
# ============================================================
# Correlation Analysis
# Calculate correlations between selected earthquake variables
# ============================================================

corr = df[[
    "magnitude",
    "depth",
    "cdi",
    "mmi",
    "sig"
]].corr()

print(corr)
```

```python
# ============================================================
# Correlation Heatmap
# Visualize the correlations between selected variables
# ============================================================

plt.figure(figsize=(8,6))

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
# ============================================================
# Depth vs Magnitude
# Visualize the relationship between earthquake depth
# and earthquake magnitude
# ============================================================

sns.scatterplot(data=df,
                x="depth",
                y="magnitude")

plt.title("Depth vs Magnitude")

plt.show()
```
